## 2.1 Credit and Debit Transactions

Credit and debit transactions are analysed to understand the movement of money through customer accounts.

A net transaction value is calculated for each transaction:

**Net Transaction = Credit Amount − Debit Amount**

- Positive values indicate that credits exceeded debits.
- Negative values indicate that debits exceeded credits.
- Values close to zero indicate relatively balanced transaction activity.

```python
#Credit and Debit Columns
df["Credit"] = df["TransactionAmount"].apply(lambda x: x if x > 0 else 0)

df["Debit"] = df["TransactionAmount"].apply(lambda x: abs(x) if x < 0 else 0)

df["NetTransaction"] = df["Credit"] - df["Debit"]

df[["TransactionAmount", "Credit", "Debit", "NetTransaction"]].head()
```

```python
df["Credit"] = df["TransactionAmount"].apply(lambda x: x if x > 0 else 0)
df["Debit"] = df["TransactionAmount"].apply(lambda x: abs(x) if x < 0 else 0)
df["NetTransaction"] = df["Credit"] - df["Debit"]
```

## 2.5 Account Activity and Transaction Gaps

Net transaction flow alone does not provide information about how frequently an account is being used.

To analyse account activity, transactions are first sorted by:

1. AccountID
2. TransactionDate

The time gap between consecutive transactions is then calculated for each account.

This allows accounts with long periods of inactivity between transactions to be identified.

### Monthly Transaction Gap

The transaction dates are converted into a month-based numerical representation to calculate the number of months between consecutive transactions for each account.

A larger MonthGap indicates a longer period without a recorded transaction.

```python

# Sort transactions by account and date
df_sorted = df.sort_values(
    ["AccountID", "TransactionDate"]
).copy()

# Convert date to datetime
df_sorted["TransactionDate"] = pd.to_datetime(
    df_sorted["TransactionDate"]
)

# Convert date into a month number
df_sorted["MonthNumber"] = (
    df_sorted["TransactionDate"].dt.year * 12
    + df_sorted["TransactionDate"].dt.month
)

# Calculate gap between consecutive transactions for each account
df_sorted["MonthGap"] = (
    df_sorted.groupby("AccountID")["MonthNumber"]
    .diff()
)

# Identify accounts with a gap of 2 months or more
dormant_accounts = (
    df_sorted[df_sorted["MonthGap"] >= 2]
    [["AccountID", "TransactionDate", "MonthGap"]]
)

dormant_accounts
```

## 2.6 Identifying Dormant / Inactive Accounts

For each account, the maximum observed gap between consecutive transactions is calculated.

Accounts with a maximum transaction gap of **3 months or more** are classified as:

**Dormant/Inactive**

Accounts with a maximum gap of less than 3 months are classified as:

**Active**

This classification is based specifically on transaction gaps observed in the dataset.

```python
account_activity = (
    df_sorted.groupby("AccountID")["MonthGap"]
    .max()
    .reset_index()
    .rename(columns={"MonthGap": "MaxGapMonths"})
)

account_activity["Status"] = account_activity["MaxGapMonths"].apply(
    lambda x: "Dormant/Inactive" if x >= 2 else "Active"
)

account_activity
```

```python
dormant_accounts_summary = account_activity[
    account_activity["Status"] == "Dormant/Inactive"
]

dormant_accounts_summary
```

### Insight

The activity analysis identifies accounts that experienced relatively long gaps between consecutive transactions.

For example, ACC10117 recorded a maximum transaction gap of **9 months**, while ACC97411 recorded a gap of **11 months**.

These prolonged gaps indicate periods of limited transaction activity and may be useful for customer engagement, account monitoring, or further risk investigation.

### Analytical Note

The term "Dormant/Inactive" is used as an analytical classification rather than an official account-status designation.

An account is classified as Dormant/Inactive when it has at least one observed gap of 2 or more months between consecutive transactions.

Therefore, this classification indicates periods of low transaction activity and should not be interpreted as confirmation of an officially dormant bank account.

## 3.2 Customer Profile Construction

To create account-level customer profiles, transaction-level data is aggregated by `AccountID`.

For each account, two key metrics are calculated:

- **Average Balance:** Mean account balance across the available transactions.
- **Transaction Volume:** Total number of transactions associated with the account.

These metrics provide a customer-level view of financial position and transaction activity.

```python
# Calculate average balance and transaction volume for each account

customer_profile = (
    df_sorted.groupby("AccountID")
    .agg(
        AverageBalance=("AccountBalance", "mean"),
        TransactionVolume=("AccountID", "size")
    )
    .reset_index()
)

customer_profile.head()
```

### Interpretation

The resulting customer profile combines balance information with transaction activity at the account level.

This allows accounts to be compared not only by how much balance they maintain, but also by how frequently they transact.

## 3.3 Defining Customer Segmentation Thresholds

### Rubric: Segment customers by average balance and transaction volume

To create objective customer segments, the median values of Average Balance and Transaction Volume are used as segmentation thresholds.

Using the median provides a data-driven method for separating accounts into relatively high and low groups without relying on arbitrary fixed thresholds.

```python
# Calculate median thresholds for balance and transaction volume

balance_median = customer_profile["AverageBalance"].median()
volume_median = customer_profile["TransactionVolume"].median()

print("Median Average Balance:", balance_median)
print("Median Transaction Volume:", volume_median)
```

### Calculated Thresholds

- **Median Average Balance:** 72,044.73
- **Median Transaction Volume:** 4.0 transactions

These thresholds are subsequently used to classify accounts into four customer segments based on their balance and transaction activity.

### Why Median Was Used

The median was selected because it represents the middle point of the account distribution and is less affected by unusually high or low account values.

Therefore, the segmentation is based on the observed characteristics of the dataset rather than an externally imposed threshold.

## 3.4 Customer Segmentation by Balance and Transaction Volume

Each account is classified using two dimensions:

- Average Balance
- Transaction Volume

The median values calculated above are used as the thresholds.

This produces four customer segments:

| Segment | Description |
|---|---|
| High Balance - High Volume | Customers maintaining relatively high balances and actively transacting |
| High Balance - Low Volume | Customers maintaining relatively high balances but conducting fewer transactions |
| Low Balance - High Volume | Customers with relatively lower balances but frequent transaction activity |
| Low Balance - Low Volume | Customers with relatively lower balances and lower transaction activity |

```python
# Segment customers based on average balance and transaction volume

def segment_customer(row):
    if row["AverageBalance"] >= balance_median and row["TransactionVolume"] >= volume_median:
        return "High Balance - High Volume"
    elif row["AverageBalance"] >= balance_median and row["TransactionVolume"] < volume_median:
        return "High Balance - Low Volume"
    elif row["AverageBalance"] < balance_median and row["TransactionVolume"] >= volume_median:
        return "Low Balance - High Volume"
    else:
        return "Low Balance - Low Volume"


customer_profile["CustomerSegment"] = customer_profile.apply(
    segment_customer,
    axis=1
)

customer_profile.head()
```

### Segment Interpretation

The segmentation provides a more detailed view of customer behavior than transaction frequency alone.

It helps distinguish between customers who are financially well-funded and active, customers who maintain balances without frequent transactions, and customers who transact frequently despite maintaining relatively lower balances.

This segmentation can support differentiated customer engagement and monitoring strategies.

## 4.1 Large Withdrawal Identification

To identify relatively large withdrawals, only transactions with a positive debit amount are considered.

A **95th percentile threshold** is calculated from the observed debit transactions. Transactions at or above this threshold are considered large withdrawals for the purpose of this analysis.

Using a percentile-based threshold allows large withdrawals to be identified relative to the transaction distribution rather than using an arbitrary fixed amount.

```python
# Consider only actual debit transactions (Debit > 0)
positive_debits = df_sorted.loc[
    df_sorted["Debit"] > 0, "Debit"
]

# Calculate the 95th percentile threshold
large_withdrawal_threshold = positive_debits.quantile(0.95)

print("Large Withdrawal Threshold:", large_withdrawal_threshold)

# Identify large withdrawal transactions
large_withdrawals = df_sorted[
    df_sorted["Debit"] >= large_withdrawal_threshold
]

# Count large withdrawals for each account
large_withdrawal_accounts = (
    large_withdrawals
    .groupby("AccountID")
    .agg(
        LargeWithdrawalCount=("Debit", "count"),
        TotalLargeWithdrawals=("Debit", "sum")
    )
    .reset_index()
    .sort_values(
        by="LargeWithdrawalCount",
        ascending=False
    )
)

large_withdrawal_accounts.head(10)
```

### Insight

The 95th percentile threshold for debit transactions is approximately **26,975.89**.

Therefore, debit transactions at or above this value represent the upper 5% of the observed debit distribution and are treated as relatively large withdrawals in this analysis.

## 4.2 Integration of Account Risk Scores

## 4.5 Withdrawal Anomaly Detection Using IQR

The **Interquartile Range (IQR)** method is used to identify unusually large withdrawal transactions.

The following measures are calculated:

- **Q1:** 25th percentile of withdrawal amounts
- **Q3:** 75th percentile of withdrawal amounts
- **IQR:** Q3 − Q1
- **Upper Limit:** Q3 + 1.5 × IQR

Withdrawal transactions above the upper limit are classified as statistical outliers.

```python
# Detect unusually large withdrawals using IQR

withdrawal_transactions = df[
    df["TransactionType"].str.lower() == "withdrawal"
].copy()

Q1 = withdrawal_transactions["TransactionAmount"].quantile(0.25)
Q3 = withdrawal_transactions["TransactionAmount"].quantile(0.75)

IQR = Q3 - Q1

upper_limit = Q3 + 1.5 * IQR

large_withdrawals = withdrawal_transactions[
    withdrawal_transactions["TransactionAmount"] > upper_limit
].copy()

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("IQR Upper Limit:", upper_limit)
print("Number of anomalous withdrawals:", len(large_withdrawals))

large_withdrawals.head(10)
```

### Insight

The IQR analysis produced an upper outlier limit of approximately **133,804.05**.

No withdrawal transactions exceeded this limit, resulting in **0 statistical outliers** according to the IQR method.

This means that the dataset does not contain withdrawal amounts extreme enough to be classified as statistical outliers using the standard IQR rule.

> **Important:** A result of zero IQR anomalies does not mean that every withdrawal is normal. It only means that no withdrawal was statistically extreme according to this specific IQR criterion.

# Task 6. Hypothesis Testing

## 6.1 Hypothesis Formulation

The objective is to determine whether high-volume transaction accounts have statistically higher average balances than low-volume transaction accounts.

### Hypotheses

**Null Hypothesis (H₀):**  
High-volume transaction accounts do not have a higher average balance than low-volume transaction accounts.

**Alternative Hypothesis (H₁):**  
High-volume transaction accounts have a statistically higher average balance than low-volume transaction accounts.

The significance level is set at **5% (α = 0.05)**.

A one-tailed Welch's independent samples t-test will be used because the two groups may have unequal variances and different sample sizes.

## 6.2 Account-Level Segmentation

To perform the hypothesis test at the account level, transaction-level data is aggregated by `AccountID`.

For each account, we calculate:

- Total number of transactions
- Average account balance

Transaction volume is measured using the number of transactions associated with each account.

```python
df.columns.tolist()
```

```python
[column for column in df.columns if "balance" in column.lower()]
```

```python
# Create account-level summary

account_summary = (
    df.groupby("AccountID")
      .agg(
          TransactionVolume=("TransactionID", "count"),
          AverageBalance=("AccountBalance", "mean")
      )
      .reset_index()
)

account_summary.head()
```

## 6.3 Volume Thresholds

To identify clearly different transaction-volume groups, the 25th and 75th percentiles of transaction volume are used.

- Accounts at or below the 25th percentile are classified as **Low Volume**.
- Accounts at or above the 75th percentile are classified as **High Volume**.
- Accounts between these thresholds are excluded from the hypothesis test.

This approach provides a clear separation between low- and high-volume accounts.

```python
# Calculate transaction volume thresholds

low_volume_threshold = account_summary["TransactionVolume"].quantile(0.25)
high_volume_threshold = account_summary["TransactionVolume"].quantile(0.75)

print("Low-volume threshold:", low_volume_threshold)
print("High-volume threshold:", high_volume_threshold)
```

## 6.4 High vs Low Volume Groups

Accounts are segmented into two groups based on the transaction-volume thresholds:

- **Low Volume:** Transaction volume ≤ 25th percentile
- **High Volume:** Transaction volume ≥ 75th percentile

The middle 50% of accounts are not included in the hypothesis test because the objective is to compare clearly low-volume and high-volume accounts.

```python
# Create low- and high-volume account groups

low_volume_accounts = account_summary[
    account_summary["TransactionVolume"] <= low_volume_threshold
].copy()

high_volume_accounts = account_summary[
    account_summary["TransactionVolume"] >= high_volume_threshold
].copy()

print("Number of low-volume accounts:", len(low_volume_accounts))
print("Number of high-volume accounts:", len(high_volume_accounts))
```

## 6.5 Average Balance Comparison

The average account balance of the high-volume and low-volume groups is compared before conducting the statistical test.

This provides an initial understanding of whether high-volume accounts appear to maintain higher average balances.

```python
# Calculate group-level average balances

low_mean_balance = low_volume_accounts["AverageBalance"].mean()
high_mean_balance = high_volume_accounts["AverageBalance"].mean()

print(f"Low-volume accounts average balance: {low_mean_balance:.2f}")
print(f"High-volume accounts average balance: {high_mean_balance:.2f}")
```

## 6.6 Welch's Independent Samples t-Test

Welch's t-test is used to determine whether the average balance of high-volume accounts is statistically greater than that of low-volume accounts.

The test is one-tailed because the research question specifically asks whether high-volume accounts have **higher** average balances.

### Significance Level

α = 0.05

### Decision Rule

- If p-value < 0.05 → Reject H₀
- If p-value ≥ 0.05 → Fail to reject H₀

```python
from scipy.stats import ttest_ind

# Remove missing balance values
low_balances = low_volume_accounts["AverageBalance"].dropna()
high_balances = high_volume_accounts["AverageBalance"].dropna()

# Welch's independent samples t-test
t_statistic, p_value = ttest_ind(
    high_balances,
    low_balances,
    equal_var=False,
    alternative="greater"
)

print(f"T-statistic: {t_statistic:.4f}")
print(f"P-value: {p_value:.6f}")
```

## 6.7 Statistical Decision

The statistical decision is based on a significance level of 5%.

If the p-value is less than 0.05, the null hypothesis is rejected, indicating statistically significant evidence that high-volume accounts have higher average balances.

If the p-value is greater than or equal to 0.05, the null hypothesis is not rejected, indicating insufficient statistical evidence to conclude that high-volume accounts have higher average balances.

```python
# Statistical decision

alpha = 0.05

if p_value < alpha:
    decision = "Reject H₀"
    conclusion = (
        "There is statistically significant evidence that "
        "high-volume accounts have higher average balances "
        "than low-volume accounts."
    )
else:
    decision = "Fail to reject H₀"
    conclusion = (
        "There is insufficient statistical evidence to conclude "
        "that high-volume accounts have higher average balances "
        "than low-volume accounts."
    )

print("Decision:", decision)
print("Conclusion:", conclusion)
```

## 6.8 Balance Distribution by Transaction Volume

A boxplot is used to compare the distribution of average account balances between high-volume and low-volume accounts.

The visualization helps identify:

- Differences in median balances
- Spread of balances
- Potential outliers
- Overall differences between the two groups

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.boxplot(
    [
        low_balances,
        high_balances
    ],
    tick_labels=["Low Volume", "High Volume"],
    patch_artist=True
)

plt.title("Average Account Balance: Low vs High Volume Accounts")
plt.xlabel("Transaction Volume Group")
plt.ylabel("Average Account Balance")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

## 6.9 Effect Size

Cohen's d is calculated to measure the magnitude of the difference in average balances between high-volume and low-volume accounts.

A commonly used interpretation is:

- |d| < 0.20 → Negligible effect
- 0.20 ≤ |d| < 0.50 → Small effect
- 0.50 ≤ |d| < 0.80 → Medium effect
- |d| ≥ 0.80 → Large effect

```python
import numpy as np

# Calculate means
high_mean = high_balances.mean()
low_mean = low_balances.mean()

# Calculate standard deviations
high_std = high_balances.std()
low_std = low_balances.std()

# Sample sizes
n_high = len(high_balances)
n_low = len(low_balances)

# Pooled standard deviation
pooled_std = np.sqrt(
    (
        (n_high - 1) * high_std**2
        + (n_low - 1) * low_std**2
    )
    / (n_high + n_low - 2)
)

# Cohen's d
cohens_d = (high_mean - low_mean) / pooled_std

print(f"Cohen's d: {cohens_d:.4f}")
```

```python
abs_d = abs(cohens_d)

if abs_d < 0.20:
    effect_interpretation = "Negligible effect"
elif abs_d < 0.50:
    effect_interpretation = "Small effect"
elif abs_d < 0.80:
    effect_interpretation = "Medium effect"
else:
    effect_interpretation = "Large effect"

print("Effect size interpretation:", effect_interpretation)
```
